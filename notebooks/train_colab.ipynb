```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
results = model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='brandsight',
)

best = results.save_dir / 'weights' / 'best.pt'
print('Best weights:', best)
```
